```python
from Linkage import linkage, euclidean_distance, plot_dendrogram,np,plt

X1 = np.array([[0.], [1.], [4.], [9.], [11.]])
Z = linkage(X1, euclidean_distance, method='single', standarize=False)
print(Z)

plot_dendrogram(Z, k=2, title="Dendrogram for X1 with single linkage")

from scipy.cluster.hierarchy import linkage as sp
print(sp(X1, method='single'))
```
